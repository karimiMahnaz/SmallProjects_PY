# using gradient decent in Regression

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.array([[1,2], [2,5], [4,7], [3,5], [5,8], [1.5,4]])
```

```python
data.T
```

```python
X_data = data.T[0]
y_data = data.T[1]
```

```python
plt.scatter(X_data, y_data)
plt.show()
```

```python
f  = lambda x,a,b: a * x + b 
```

```python
a = np.random.random()
```

```python
b = np.random.random()
```

```python
x = np.arange(0,10)
y_plot = [f(i,a,b) for i in x]
plt.scatter(X_data, y_data)
plt.plot(x, y_plot, color='red')
plt.show()
```

```python
y_eval = [f(x,a,b) for x in X_data]
```

```python
# MSE : 1/n(sum(ypred - ytrue) ** 2)

mse_eval =  np.square(y_eval - y_data).mean()
```

```python
delta = .01
```

```python
delta_a = a + delta
```

```python
y_delta_a = [f(x, delta_a, b) for x in X_data]
```

```python
mse_delta_a = np.square(y_delta_a - y_data).mean()
```

```python
print(mse_delta_a)
print(mse_delta_a ) 
print(mse_eval)
```

```python
delat_mse_a = mse_delta_a - mse_eval
```

```python
delat_mse_a
```

```python
m_a = delat_mse_a / delta
```

```python
m_a
```

```python
lr=.01
```

```python
a = a - lr * m_a
```

```python
epoch = 100
iteration = 1
lr_a = .01
lr_b = .001
mse =[]

while iteration <= epoch:

    y_eval = [f(x,a,b) for x in X_data]
    mse_eval = np.square(y_eval - y_data).mean()
    mse.append(mse_eval)
    delta = .01
    delta_a = a + delta
    delta_b = b + delta

    # print(delta_a)
    # print(delta_b)

    y_delta_a = [f(x, delta_a, b) for x in X_data] 
    y_delta_b = [f(x, a, delta_b) for x in X_data] 

    mse_delta_a = np.square(y_delta_a - y_data).mean()
    mse_delta_b = np.square(y_delta_b - y_data).mean()
    
    delat_mse_a = mse_delta_a - mse_eval
    delat_mse_b = mse_delta_b - mse_eval

    m_a = delat_mse_a / delta
    m_b = delat_mse_b / delta
    
    a = a - lr_a * m_a
    b = b - lr_b * m_b
    

    # print(m_a)
    # print(m_b)

   
    print(f"epoch: {iteration}, error: {mse_eval} ")
    iteration+=1
```
