```python
text = '؟می باشد دیتایاد - مرجع تخصصی ،  علم داده و هوش مصنوعی'
```

```python
import hazm
```

```python
normalizer = hazm.Normalizer()
text = normalizer.normalize(text)
```

```python
print(text)
```

```python
tokenizer = hazm.WordTokenizer()
tokens = tokenizer.tokenize(text)
```

```python
stop_word = hazm.stopwords_list()
```

```python
sentence = []
for token in tokens:
    if token not in stop_word:
        sentence.append(token)
```
