```python
def text_cleaner(text):
    normalizer = hazm.Normalizer()
    text = normalizer.normalize(text)

    tokenizer = hazm.WordTokenizer()
    tokens = tokenizer.tokenize(text)
    
    stop_word = hazm.stopwords_list()
    punctuation = list(string.punctuation + '؟،؛')
    stemmer = hazm.Stemmer()

    filter_sentence = []
    for token in tokens:
        if (not token.isdigit()) and (token not in stop_word) and (token not in punctuation):
            filter_sentence.append(stemmer.stem(token))
    
    text = ' '.join(filter_sentence)
        
    return text
    
```

```python
%%time
for i in range(10):
    text = train_data['content'][i]
    text_c = text_cleaner(text)
```

```python
train_data['content'] = train_data['content'].apply(text_cleaner)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
```

```python
vectorizer = TfidfVectorizer(min_df=2)
X_train = vectorizer.fit_transform(train_data['content'])
```

```python
X_train.shape
```

```python
svm_clf = SVC()
svm_clf.fit(X_train,y_train)
```

```python
svm_clf.score(X_train,y_train)
```
