```python
text_science = '''
        Scientists discover a new species of the marine life in deep ocean.
        The discovery is a significant advancement in marine biology are And and and.
                '''
```

```python
text_economy = '''
        Stock markets soar as companies report record profits for the quarter.
        Investors are optimistic about the future of the economy.
                '''
```

```python
text_sport = '''
        Team USA clinches gold in a thrilling overtime victory at the Winter Olympics,
        showcasing their determination and skill on the ice.
            '''
```

```python
documents = [text_science,text_economy,text_sport]
labels = ['science','economy','sport']
```

```python
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
```

```python
vectorizer = CountVectorizer(ngram_range=(1,2),lowercase=True)
bow = vectorizer.fit_transform(documents)
```

```python
X = bow.toarray()
```

```python
X.shape
```

```python
X.shape
```

```python
features = vectorizer.get_feature_names_out()
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(X,columns=features)
```

```python
#tf-idf
#tf: term frequency
#idf : inverse document frequency
```

```python
tfidf_vectorizer = TfidfVectorizer()
X_tfidf = tfidf_vectorizer.fit_transform(documents)
```

```python
features = tfidf_vectorizer.get_feature_names_out()
```

```python
df = pd.DataFrame(X_tfidf.toarray(),columns=features)
```
