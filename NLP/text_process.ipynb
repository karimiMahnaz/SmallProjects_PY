```python
# text processing
```

```python
text_science = '''
        Scientists discover a new species of the marine life in deep ocean.
        The discovery is a significant advancement in marine biology are And and and.
                '''
```

```python
text_economy = '''
        Stock markets soar as companies report record profits for the quarter.
        Investors are optimistic about the future of the economy.
                '''
```

```python
text_sport = '''
        Team USA clinches gold in a thrilling overtime victory at the Winter Olympics,
        showcasing their determination and skill on the ice.
            '''
```

```python
token_text_science = word_tokenize(text_science)
token_text_economy = word_tokenize(text_economy)
token_text_sport = word_tokenize(text_sport)
```

```python
stemmer = PorterStemmer()
filtered_sentence_science = []
text_science = ''
for token in token_text_science:
    if (token not in stop_words) and (token not in punctuation):
        word = stemmer.stem(token)
        filtered_sentence_science.append(word)

stemmer = PorterStemmer()
filtered_sentence_economy = []
text_economy = ''
for token in token_text_economy:
    if (token not in stop_words) and (token not in punctuation):
        word = stemmer.stem(token)
        filtered_sentence_economy.append(word)

stemmer = PorterStemmer()
filtered_sentence_sport = []
text_sport = ''
for token in token_text_sport:
    if (token not in stop_words) and (token not in punctuation):
        word = stemmer.stem(token)
        filtered_sentence_sport.append(word)
```
