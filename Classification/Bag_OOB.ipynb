```python
def oob_prob(n):
    return (1-1/n)**n
```

```python
probs = []
for n in range(1, 101):
    prob = oob_prob(n)
    probs.append(prob)
```

```python
plt.figure(figsize=(15, 10))
plt.plot(probs)
plt.hlines(.3669, 0, 110, colors="orange", linestyles="--")
plt.show()
```
