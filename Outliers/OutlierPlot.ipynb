Boxplot- Zcore - MAD

```python
import numpy as np
```

```python
data = np.array([ [1, 2, 5],
                 [3, 10, 1 ],
                 [10 , -10, 0],
                 [2, 50, 30],
                 [21, 21, -10],
                 [0, -50, 4],
                 [60, 22, 40]
])
```

```python
Q1 = np.percentile(data, 25, axis = 0)
Q3 = np.percentile(data, 75, axis = 0)

IQR = Q3 - Q1
```

```python
lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)
```

```python
(data > upper_bound) | (data < lower_bound)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.boxplot(data)
plt.show()
```

```python
sns.boxplot(data)
plt.show()
```

Zscore : can use Zscore for Normal distribution

```python
# Zscore for Normal distribution
```

```python

np.random.seed(123456)
data0 = np.random.normal(loc = 3 , scale = 7, size = [10,3])
```

```python
from scipy.stats import zscore
```

```python
z_scores = zscore(data0, axis = 0)
```

```python
z_threshold = 2
```

```python
np.abs(z_scores) > z_threshold
```

MAD : Median Absolute Deviation
for non normal distribution

```python
medians = np.median(data0, axis =0)
```

```python
medians
```

```python
MAD = np.median(np.abs(data0 - medians), axis = 0)
MAD
```
