OCSVM: One Class SVM
unsupervised model
label: oulier = -1 and inlier = 1


SVM Objective Function : L = 1/2||w|| + c * sum(ei)

One Class SVM Objective Function : 1/2||w|| + 1/n*nu sum(ei) - r

with bigger r we have few outliers 


```python
import numpy as np
import pandas as  pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("data.csv")
```

```python
data.head()
```

```python
data.isna().sum()
```

```python
X = data.drop(columns='class')
y = data['class']
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report
```

```python
scalar = StandardScaler()
X = scalar.fit_transform(X)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
for col in data.columns:
    if data[col].dtype == 'object':
        le = LabelEncoder()
        data[col] = le.fit_transform(data[col])
```

```python
data['class'][data['class'] == 0 ] = -1
```

```python
# OR
data.iloc[: , :-1] = scalar.fit_transform(data.iloc[:, :-1])
```

```python
from sklearn.svm import OneClassSVM
```

```python
model = OneClassSVM(nu=.1)
model.fit(X)
```

```python
y_pred = model.predict(X)
```

```python
np.unique(y_pred, return_counts=True)
```

```python
print(classification_report(y, y_pred))
```

```python
data_inlier = data[data['class'] == 1] 
data_outlier = data[data['class'] == -1] 
```

```python
data_test = pd.concat(data_inlier[:2000], data_outlier[:2000])
data_train = data_inlier[1500:]
```

```python
x_train = data_train.drop(columns='class')
y_train = data_train['class']

x_test = data_test.drop(columns='class')
y_test = data_test['class']
```

```python
model = OneClassSVM(nu=.06)
model.fit(x_train)
```

```python
y_pred_train = model.predict(x_train)
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
y_pred_test = model.predict(x_test)
```

```python
np.unique(y_pred_test, return_counts=True)
```

```python
print(classification_report(y_test,y_pred_test))
```
