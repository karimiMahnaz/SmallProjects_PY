LOF : Local Outlier Factor

Density has Inverse relationship with average distance from neighbors

1 - find neighbors
2 - Density of points -> LRD = n/sum(d(x,y))
3 - LOF of points and its neighbors ->
    LOF(a) = ((LRD(b) + LRD(c))/2) / LRD(a)

   if LOf = 1 -> it has same density with neighbors
   if LOF > 1 -> neighbors' density is more (point is outlier)
   if LOF < 1 -> it's density is more than neighbors ( point is inlier)

   labele: oulier = -1 and inlier = 1

```python
import numpy as np
import pandas as  pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("data.csv")
```

```python
data.head()
```

```python
data.isna().sum()
```

```python
X = data.drop(columns='class')
y = data['class']
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report
```

```python
scalar = StandardScaler()
X = scalar.fit_transform(X)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
for col in data.columns:
    if data[col].dtype == 'object':
        le = LabelEncoder()
        data[col] = le.fit_transform(data[col])
```

```python
data['class'][data['class'] == 0 ] = -1
```

```python
# OR
data.iloc[: , :-1] = scalar.fit_transform(data.iloc[:, :-1])
```

```python
from sklearn.neighbors import LocalOutlierFactor
```

```python
lof = LocalOutlierFactor(n_neighbors=10, novelty=True, contamination= .45)

# y_pred = lof.fit_predict(X)  # when novelty=False

lof.fit(X)  # when novelty=True
y_pred = lof.predict(X)  # when novelty=True
```

```python
print(classification_report(y,y_pred))
```

```python
np.unique(y_pred, return_counts=True)
```

 in above solution we don't have acceptable result
 in below we will have train set with inliers (normal data)

```python
data_inlier = data[data['class'] == 1] 
data_outlier = data[data['class'] == -1] 
```

```python
data_test = pd.concat(data_inlier[:2000], data_outlier[:2000])
data_train = data_inlier[1500:]
```

```python
x_train = data_train.drop(columns='class')
y_train = data_train['class']

x_test = data_test.drop(columns='class')
y_test = data_test['class']
```

```python
lof = LocalOutlierFactor(n_neighbors=10, novelty=True, contamination= .45)

# y_pred = lof.fit_predict(X)  # when novelty=False

lof.fit(X)  # when novelty=True
y_pred = lof.predict(X)  # when novelty=True
```

```python
y_pred_train = lof.predict(x_train)
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
y_pred_test = lof.predict(x_test)
```

```python
np.unique(y_pred_test, return_counts=True)
```

```python
print(classification_report(y_test,y_pred_test))
```
