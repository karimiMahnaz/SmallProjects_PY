```python
from tensorflow.keras.layers import Dense, Conv2D, Dropout, Input, Concatenate, Flatten, MaxPool2D
from tensorflow.keras.models import Model
```

```python
input_layer = Input(shape=(28,28,1))

conv_layer1 = Conv2D(16,(3,3))(input_layer)
conv_layer2 = Conv2D(32,(3,3))(conv_layer1)

flatten = Flatten()(conv_layer2)

dense_layer1 = Dense(128)(flatten)

output_layer = Dense(10)(dense_layer1)
```

```python
model = Model(inputs = input_layer, outputs = output_layer)
```

```python
model.summary()
```

```python
# multi channel - multi head

input_layer_ch1 = Input(shape = (24,))
input_layer_ch2 = Input(shape = (34,))

concat = Concatenate()([input_layer_ch1, input_layer_ch2])

output_layer= Dense(10)(concat)

```

```python
model = Model(inputs = [input_layer_ch1, input_layer_ch2], outputs = output_layer)
```

```python
model.summary()
```

```python
# inception module

input_layer = Input(shape = (100,100,3))

conv_layer_1x1 = Conv2D (5,(1,1), padding='same')(input_layer)
conv_layer_3x3 = Conv2D (16,(3,3), padding='same')(input_layer)
conv_layer_5x5 = Conv2D (16,(3,3), padding='same')(input_layer)
max_pool_3x3 = MaxPool2D(pool_size=(3,3), strides=(1,1), padding='same')(input_layer)

concat = Concatenate()([conv_layer_1x1, conv_layer_3x3, conv_layer_5x5, max_pool_3x3])

output_layer = Dense(10)(concat)
```

```python
model_inception = Model(inputs = input_layer, outputs = output_layer)
```

```python
model_inception.summary()
```

```python
from tensorflow.keras.utils import plot_model
```

```python
plot_model(model_inception, show_shapes=True)
```
