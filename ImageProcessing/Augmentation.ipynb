```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img, img_to_array
```

```python

image = load_img("cat.jpg")
image = img_to_array(image, dtype="uint8")
```

```python
image = image.reshape(1, 3803, 2853, 3 )
```

```python
# data_gen = ImageDataGenerator(rescale=1./255, zoom_range=.05, rotation_range=10, dtype='uint8')

data_gen = ImageDataGenerator(zoom_range=.05, rotation_range=10, dtype='uint8')
```

```python
image_gen = data_gen.flow(image, batch_size = 32)
```
