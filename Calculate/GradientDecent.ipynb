# Gradient Decent find a local point

```python
f = lambda x: (x+2) ** 2 + 1
fd = lambda x: 2*x + 4

x0 = 2
```

```python
x1 = x0 - fd(x0)
```

```python
x1
```

```python
eta= .1
```

```python
x1 = x0 - eta * fd(x0)
x1
```

```python
x2 = x1 - eta * fd(x1)
x2
```

```python
x3 = x2 - eta * fd(x2)
x3
```

```python
x0 = 2
#learning rate
eta = .1
epoch = 100
iterate =1
values = [[],[]]

while epoch > 0:
    values[0].append(x0)
    values[1].append(f(x0))
    
    x = x0 -(eta * fd(x0))
    delta = f(x)-f(x0)
    print(f"epoch : {iterate}, error: {f(x0)}, delta: {delta}")

    if abs(delta) < .001:
        break
    
    x0 = x
    epoch -= 1
    iterate +=1
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.arange(-5, 4, .1)
y = [f(i) for i in x]

plt.plot(x,y)
plt.plot(values[0], values[1], marker = 'o')

plt.show
```
