```python
def integral_x2(start,stop,split):
    delta_x = (stop - start) / split
    area = 0
    
    while start < stop:
        a = start
        start += delta_x
        area = area + delta_x * (a**2)
    
    return area

A = 2.3333
```

```python
integral_x2(1,2,5) - A
```

```python
integral_x2(1,2,10) - A
```

```python
integral_x2(1,2,100) - A
```

```python
integral_x2(1,2,1000) - A
```

```python
integral_x2(1,2,10000000)
```
