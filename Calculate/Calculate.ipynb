```python
import sympy as sp
from sympy import symbols, diff, sin, lambdify
```

```python
x, y = symbols('x y')
f = sin(x*y)
df_dx = diff(f, x)  # Partial derivative

# Method 1: Substitute and evaluate
value = df_dx.subs({x: 1, y: 2}).evalf()
print(value)

# Method 2: Lambdify
df_numeric = lambdify((x, y), df_dx)
print(df_numeric(1, 2))  # Evaluates at x=1, y=2
```

```python
x = symbols('x')
f0 = x**2 + 10
df = diff(f0,x)
#df.doit_numerically(8)


# Convert to a numerical function
df_num = lambdify(x, df)  # Now df_num(8) computes 2*8 = 16

# Evaluate at x=8
result = df_num(8)  # Output: 16
print(result)
```

```python

x = symbols('x')
f0 = x**2 + 10

# Create an unevaluated derivative
df = sp.Derivative(f0, x)  # Keeps it as Derivative(x**2 + 10, x)
df.doit()
```

```python
# Now doit_numerically() works
result = df.doit_numerically(8)  # Output: 16.0
print(result)
```
