```python
import sympy as sp
```

```python
x = sp.Symbol('x')
y = sp.Symbol('y')
```

```python
f = x**2  + 10
```

```python
f
```

```python
sp.diff(f,y)
```

```python
d = sp.diff(f)
```

```python
sp.solve(d)
```

```python
d
```

```python
f
```

```python
deriv = sp.Derivative(f,x)
```

```python
deriv
```

```python
deriv.doit()
```

```python
deriv.doit_numerically(10)
```

```python
sp.integrate(f)
```
